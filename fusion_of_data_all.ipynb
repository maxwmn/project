# 1.0 Fusion of all data
**1.1 Fusion of the data belonging to the wind turbine with asset_id = 0**

```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#the datasets 0, 24, 26, 71, and 73 are the datasets for the turbine with asset_id = 0.

df0 = pd.read_csv("Wind Farm A\\datasets\\0.csv", parse_dates=True, sep=";")
df24 = pd.read_csv("Wind Farm A\\datasets\\24.csv", parse_dates=True, sep=";")
df26 = pd.read_csv("Wind Farm A\\datasets\\26.csv", parse_dates=True, sep=";")
df71 = pd.read_csv("Wind Farm A\\datasets\\71.csv", parse_dates=True, sep=";")
df73 = pd.read_csv("Wind Farm A\\datasets\\73.csv", parse_dates=True, sep=";")

#label the datasets with their respective event_id
df0_labeled = df0.copy()
df0_labeled['event_id'] = 0
df24_labeled = df24.copy()
df24_labeled['event_id'] = 24
df26_labeled = df26.copy()
df26_labeled['event_id'] = 26
df71_labeled = df71.copy()
df71_labeled['event_id'] = 71
df73_labeled = df73.copy()
df73_labeled['event_id'] = 73

df_t0 = pd.concat([df0_labeled, df24_labeled, df26_labeled, df71_labeled, df73_labeled], ignore_index=True)

df_t0.to_csv("Wind Farm A\\datasets\\turbine_data\\t0.csv", index=False, sep=";")

print("Shape:",df_t0.shape)
print("Head:")
print(df_t0.head())
print("Tail:")
print(df_t0.tail())
```
