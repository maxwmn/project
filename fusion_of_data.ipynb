# 1.0 Fusion of all data
**1.1 Fusion of the data belonging to the wind turbine with asset_id = 0**

```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#the datasets 0, 24, 26, 71, and 73 are the datasets for the turbine with asset_id = 0.

df0 = pd.read_csv("Wind Farm A\\datasets\\0.csv", parse_dates=True, sep=";")
df24 = pd.read_csv("Wind Farm A\\datasets\\24.csv", parse_dates=True, sep=";")
df26 = pd.read_csv("Wind Farm A\\datasets\\26.csv", parse_dates=True, sep=";")
df71 = pd.read_csv("Wind Farm A\\datasets\\71.csv", parse_dates=True, sep=";")
df73 = pd.read_csv("Wind Farm A\\datasets\\73.csv", parse_dates=True, sep=";")
df_t0 = pd.concat([df0, df24, df26, df71, df73], ignore_index=True)


df_t0_sorted = df_t0.sort_values(by="time_stamp")
df_t0_sorted.to_csv("Wind Farm A\\datasets\\t0_sorted.csv", index=False, sep=";")
df_t0.to_csv("Wind Farm A\\datasets\\t0.csv", index=False, sep=";")

print("Shape:",df_t0.shape)
print("First 5 rows:")
print(df_t0_sorted.head())
print(df_t0_sorted.tail())
```
